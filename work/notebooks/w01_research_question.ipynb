## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*


The starter dataset contains 30,000 content pages.

Among these pages, 13,152 pages (43.84%) were declining while still receiving at least 100 impressions. I also found 9,759 visible pages (32.53%) with at least 500 impressions, an average position between 1 and 20, and CTR below 0.5%.

In addition, 7,076 pages (23.59%) were at least 180 days old while still appearing within the top 10 average positions.

These observed groups suggest that the dataset contains enough potentially valuable review candidates to investigate a ranked refresh-opportunity system over the next seven weeks.


```python
import os
import pandas as pd

while not os.path.isdir("data/raw") and os.getcwd() != os.path.dirname(os.getcwd()):
    os.chdir("..")

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)

declining_with_demand = (
    (df["trend_direction"] == "down")
    & (df["impressions_90d"] >= 100)
).sum()

low_ctr_visible = (
    (df["impressions_90d"] >= 500)
    & (df["avg_position"] > 0)
    & (df["avg_position"] <= 20)
    & (df["ctr"] < 0.5)
).sum()

page_one_age_risk = (
    (df["avg_position"] > 0)
    & (df["avg_position"] <= 10)
    & (df["content_age_days"] >= 180)
).sum()

summary = pd.DataFrame(
    {
        "Metric": [
            "Total pages",
            "Declining pages with at least 100 impressions",
            "Visible low-CTR pages",
            "Older pages positioned in the top 10",
        ],
        "Count": [
            total_pages,
            declining_with_demand,
            low_ctr_visible,
            page_one_age_risk,
        ],
    }
)

summary["Percentage of pages"] = (summary["Count"] / total_pages * 100).round(2)
summary
```
